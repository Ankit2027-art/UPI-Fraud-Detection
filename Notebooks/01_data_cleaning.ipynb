Step 8 - Convert Timestamp data type Onject To datetime

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
print(df["Timestamp"].dtype)
```

Step 9 -Amount Validation

```python
df["Amount"] =pd.to_numeric(df["Amount"],errors ="coerce")
```

```python
print(df["Amount"].describe())
```

Negative

```python
print("Negative Amounts:",(df["Amount"] < 0).sum())
```

Zero

```python
print("Zero amounts:",(df["Amount"] == 0).sum())
```

Step 10 - Categorical Values Cheak

```python
categorical_columns = [
    "Merchant_Category",
    "Transaction_Type",
    "Location",
    "Device_Type",
    "UPI_Channel",
    "Day_of_Week"
]
for col in categorical_columns:
    print(df[col].unique())
```

```python
for col in categorical_columns:
    print(df[col].value_counts())
```

Step 11 - ID Validation

```python
print("Unique Sender Ids:",df["Sender_ID"].nunique())
print("Unique Reciver IDs:",df["Receiver_ID"].nunique())
```

Missing IDs

```python
print("Missing Sender IDs:",df["Sender_ID"].isna().sum())
print("Missing Reciver IDs:",df["Receiver_ID"].isna().sum())
```

Step 12 -Time Features Validation

```python
df["Hour_of_Day"] =df["Timestamp"].dt.hour
df["Day_of_Week"] =df["Timestamp"].dt.day_name()
```

```python
df["Is_Weekend"] = (df["Timestamp"].dt.dayofweek >= 5).astype(int)
```

```python
df[[
    "Timestamp",
    "Hour_of_Day",
    "Day_of_Week",
    "Is_Weekend"
]].head()
```
